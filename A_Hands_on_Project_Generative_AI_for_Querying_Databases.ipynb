## Chest Pain Type Frequency

Consider the following prompt.

**Prompt**

```html
Write an SQL query to determine the frequency of each type of chest pain (typical angina, atypical angina, non-anginal pain, asymptomatic) among patients.
```

We can expect the following query to be generated.


````python
query = """
SELECT
    CASE
        WHEN cp = 1 THEN 'Typical angina'
        WHEN cp = 2 THEN 'Atypical angina'
        WHEN cp = 3 THEN 'Non-anginal pain'
        WHEN cp = 4 THEN 'Asymptomatic'
    END AS chest_pain_type,
    COUNT(*) AS frequency
FROM heart_disease
GROUP BY cp
ORDER BY cp;
"""
result = pd.read_sql_query(query, conn)
display(result)
````

### Age Group Analysis and Target Variable

Investigate the distribution of the target variable (presence or absence of heart disease) across different age groups (e.g., 20–30, 30–40, etc.).

````python
query = """
SELECT
    CASE
        WHEN age >= 20 AND age < 30 THEN '20-29'
        WHEN age >= 30 AND age < 40 THEN '30-39'
        WHEN age >= 40 AND age < 50 THEN '40-49'
        WHEN age >= 50 AND age < 60 THEN '50-59'
        WHEN age >= 60 AND age < 70 THEN '60-69'
        WHEN age >= 70 THEN '70+'
    END AS age_group,
    CASE
        WHEN num = 0 THEN 'Absence of heart disease'
        WHEN num > 0 THEN 'Presence of heart disease'
    END AS heart_disease_status,
    COUNT(*) AS patient_count
FROM heart_disease
GROUP BY age_group, heart_disease_status
ORDER BY age_group, heart_disease_status;
"""

result = pd.read_sql_query(query, conn)
display(result)
````

 ### Maximum Heart Rate by Age Group

Find the maximum heart rate achieved during exercise for different age groups (e.g., 30–40, 40–50, etc.).

````python
query = """
SELECT
    CASE
        WHEN age >= 30 AND age < 40 THEN '30-39'
        WHEN age >= 40 AND age < 50 THEN '40-49'
        WHEN age >= 50 AND age < 60 THEN '50-59'
        WHEN age >= 60 AND age < 70 THEN '60-69'
        WHEN age >= 70 THEN '70+'
    END AS age_group,
    MAX(thalach) AS max_heart_rate
FROM heart_disease
WHERE age >= 30
GROUP BY age_group
ORDER BY MIN(age);
"""

result = pd.read_sql_query(query, conn)
display(result)
````

### Percentage of Patients with High Blood Sugar

Calculate the percentage of patients with fasting blood sugar greater than 120 mg/dL

````python
query = """
SELECT
    COUNT(CASE WHEN fbs = 1 THEN 1 END) * 100.0 / COUNT(*)
        AS percentage_above_120
FROM heart_disease;
"""

result = pd.read_sql_query(query, conn)
display(result)
````

### Ratio of Patients with Resting Electrocardiographic Abnormalities

Find the ratio of patients with abnormal resting electrocardiographic results to those with normal results.

````python
query = """
SELECT
    SUM(CASE WHEN restecg IN (1, 2) THEN 1 ELSE 0 END)
        AS abnormal_ecg_patients,
    SUM(CASE WHEN restecg = 0 THEN 1 ELSE 0 END)
        AS normal_ecg_patients,
    ROUND(
        CAST(SUM(CASE WHEN restecg IN (1, 2) THEN 1 ELSE 0 END) AS FLOAT)
        / NULLIF(SUM(CASE WHEN restecg = 0 THEN 1 ELSE 0 END), 0),
        2
    ) AS abnormal_to_normal_ratio
FROM heart_disease;
"""

result = pd.read_sql_query(query, conn)
display(result)
````

### Number of Patients with Reversible Thalassemia
